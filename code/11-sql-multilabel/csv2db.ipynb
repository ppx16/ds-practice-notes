```python
import pandas as pd
import sqlite3

# 連接 SQLite 資料庫（如果不存在則會自動建立）
conn = sqlite3.connect("alldata.db")

# 讀取 CSV 檔案
questions_df = pd.read_csv("Filtered_Questions.csv")
answers_df = pd.read_csv("Filtered_Answers.csv")
test_df = pd.read_csv("Test_data.csv")
train_df = pd.read_csv("Train_data.csv")

# 將數據寫入 SQLite 資料庫
questions_df.to_sql("Filtered_Questions", conn, if_exists="replace", index=False)
answers_df.to_sql("Filtered_Answers", conn, if_exists="replace", index=False)
test_df.to_sql("Test_data", conn, if_exists="replace", index=False)
train_df.to_sql("Train_data", conn, if_exists="replace", index=False)

# 關閉資料庫連線
conn.close()

print("所有數據已成功寫入 SQLite 資料庫 'alldata.db'")
```

```python
import sqlite3
import pandas as pd

# 連接 SQLite 資料庫
conn = sqlite3.connect("alldata.db")

# 取得所有表格名稱
tables = ["Filtered_Questions", "Filtered_Answers", "Test_data", "Train_data"]

# 檢查每個表的欄位數量與數據筆數
for table in tables:
    # 讀取表格資料
    df = pd.read_sql(f"SELECT * FROM {table}", conn)
    print(df.head(5))
    print(df['Id'].nunique())
    
    # 取得欄位數量
    num_columns = len(df.columns)
    
    # 取得數據筆數
    num_rows = len(df)
    
    print(f"表格 `{table}` 有 {num_columns} 個欄位，{num_rows} 筆數據")

# 關閉資料庫連線
conn.close()
```

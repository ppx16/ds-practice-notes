```python
import pandas as pd
from sklearn.model_selection import train_test_split

# Step 1: 載入資料集
data = pd.read_csv("fraud_indicators.csv")

# Step 2: 檢查FraudIndicator欄位
if 'FraudIndicator' not in data.columns:
    raise ValueError("資料集中未找到 'FraudIndicator' 欄位")

# Step 3: 分層隨機抽樣
train_data, test_data = train_test_split(
    data,
    test_size=0.2,        # 測試資料占比
    stratify=data['FraudIndicator'],  # 分層依據
    random_state=42      # 設定隨機種子以重現結果
)

# Step 4: 確認分層效果
print("訓練資料 FraudIndicator 分布：")
print(train_data['FraudIndicator'].value_counts(normalize=True))

print("\n測試資料 FraudIndicator 分布：")
print(test_data['FraudIndicator'].value_counts(normalize=True))

# Step 5: 儲存分割後的資料（如果需要）
train_data.to_csv("train_data.csv", index=False)
test_data.to_csv("test_data.csv", index=False)
```

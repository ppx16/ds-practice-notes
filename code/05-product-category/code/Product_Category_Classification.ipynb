```python
import pandas as pd
import numpy as np
```

```python
df_train = pd.read_excel('../data/training.xlsx')
df_test = pd.read_excel('../data/test-Ans.xlsx')
```

### 数据预处理

```python
list_columns = df_train['Product_Category'].unique()
list_columns
```

```python
list_nums = np.arange(len(df_train['Product_Category'].unique()))
list_nums
```

```python
cate_dict = dict()

for key,value in zip(list_columns, list_nums):
    cate_dict.update({key:value})
    
cate_dict
```

```python
num_dict = dict()

for key,value in zip(list_nums, list_columns):
    num_dict.update({key:value})
    
num_dict
```

```python
df_train['target'] = df_train['Product_Category'].apply(lambda x :cate_dict.get(x))
df_train.head(5)
```

```python
df_test['target'] = df_test['Product_Category'].apply(lambda x :cate_dict.get(x))
df_test.head(5)
```

```python
sum(df_train['Product_Description'].isna())
```

```python
df_train['Product_Description']=df_train['Product_Description'].fillna('')
df_test['Product_Description']=df_test['Product_Description'].fillna('')
```

### 加载停用词

```python
stopwords = set()

with open('stopwords.txt','r',encoding='utf-8') as sw:
    [stopwords.add(line.strip()) for line in sw.readlines()] #list Comprehension
    
stopwords.remove('')
len(stopwords)
```

### 分词 & 词性标注

```python
import jieba
import jieba.posseg as pseg

def des_clean(s):
    #segs = jieba.cut(s, cut_all=False)
    #print u"[精确模式]: ", "  ".join(segs)
    #分词并标注词性
    segs = pseg.cut(s)
    
    final = ''
    
    for seg ,flag in segs:
        #去停用词
        if seg not in stopwords:
           #去数词和去字符串
           if flag !='m' and flag != 'x' and flag != 'ul':
                #输出分词
                final +=' '+ seg
    return final
```

```python
df_train['Product_Description'].iloc[0], des_clean(df_train['Product_Description'].iloc[0])
```

```python
df_train['Product_Description_clean']=df_train['Product_Description'].apply(lambda x :des_clean(x))
df_test['Product_Description_clean']=df_test['Product_Description'].apply(lambda x :des_clean(x))
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(df_train.Product_Description_clean, df_train.target, test_size=0.25, random_state=33)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tv = TfidfVectorizer(sublinear_tf=True, decode_error='ignore')
x_train = tv.fit_transform(X_train)
x_test = tv.transform(X_test)

x_train.shape, x_test.shape
```

### 建模

### 1. 朴素贝叶斯

```python
from sklearn.model_selection import GridSearchCV
from sklearn.naive_bayes import MultinomialNB

mnb = MultinomialNB()
MNB_params = {'alpha':[0.02,0.05, 0.1,0.5]} #拉普拉斯平滑算子
MNB_grid = GridSearchCV(mnb, MNB_params, cv = 5, verbose = 1, scoring = 'neg_log_loss', n_jobs = -1)
MNB_grid.fit(x_train, y_train)
MNB_grid.best_estimator_
```

```python
from sklearn.metrics import accuracy_score, f1_score

y_pred = MNB_grid.predict(x_test)

print('Accuracy:', accuracy_score(y_test, y_pred))
print('Micro F1:', f1_score(y_test, y_pred, average='micro'))
```

```python
from sklearn.metrics import classification_report

print(classification_report(y_test, y_pred))
```

### 2. 随机森林

```python
from imblearn.over_sampling import SMOTE

sm = SMOTE()

x_new, y_new = sm.fit_resample(x_train, y_train)
```

```python
from sklearn.ensemble import RandomForestClassifier

rf = RandomForestClassifier(n_estimators = 1000, n_jobs=-1).fit(x_new, y_new)

y_pred = rf.predict(x_test)

print('Accuracy:', accuracy_score(y_test, y_pred))
print('Micro F1:', f1_score(y_test, y_pred, average='micro'))
```

```python
print(classification_report(y_test, y_pred))
```

```python
rf = RandomForestClassifier(class_weight = 'balanced', n_estimators = 1000, n_jobs = -1).fit(x_train, y_train)

y_pred = rf.predict(x_test)

print('Accuracy:', accuracy_score(y_test, y_pred))
print('Micro F1:', f1_score(y_test, y_pred, average='micro'))
```

```python
print(classification_report(y_test, y_pred))
```

### 3. XGBOOST

```python
from xgboost import XGBClassifier

xgb = XGBClassifier(n_estimators=360, max_depth=3, learning_rate=0.1).fit(x_new, y_new)

y_pred = xgb.predict(x_test)

print('Accuracy:', accuracy_score(y_test, y_pred))
print('Micro F1:', f1_score(y_test, y_pred, average='micro'))
```

```python
print(classification_report(y_test, y_pred))
```

### 产生最终输出结果

```python
X_train = df_train.Product_Description_clean
y_train = df_train.target
X_test = df_test.Product_Description_clean
y_test = df_test.target
```

```python
tv = TfidfVectorizer(sublinear_tf=True, decode_error='ignore')
x_train = tv.fit_transform(X_train)
x_test = tv.transform(X_test)

x_train.shape, x_test.shape
```

```python
mnb = MultinomialNB()
MNB_params = {'alpha':[0.02,0.05, 0.1,0.5]} #拉普拉斯平滑算子
MNB_grid = GridSearchCV(mnb, MNB_params, cv = 5, verbose = 1, scoring = 'neg_log_loss', n_jobs = -1)
MNB_grid.fit(x_train, y_train)
y_pred = MNB_grid.predict(x_test)
```

```python
df_test['label'] = y_pred
df_test.head(5)
```

```python
df_test['Predicted_Product_Category'] = df_test['label'].apply(lambda x :num_dict.get(x))
df_test.head(5)
```

```python
df_test[['Product_ID', 'Predicted_Product_Category']].to_csv('Final_Results.csv', encoding='gb18030', index = False)
```
